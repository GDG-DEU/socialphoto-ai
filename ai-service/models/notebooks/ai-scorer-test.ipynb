 Aİ SCORER MODELİNİN UYGUN OLUP OLMADIĞININ ANLAŞILMASI İÇİN TEST 4 NORMAL GÖRSEL + 1 SİYAH GÖRSEL

```python
from transformers import CLIPProcessor, CLIPModel
from aesthetic_scorer import AestheticScorer
import torch
from PIL import Image


image1 = Image.open("1.jpg")
image2 = Image.open("2.jpg")
image3 = Image.open("3.jpg")
image4 = Image.open("4.jpg")
image_black = Image.new('RGB', (224, 224), color='black')  # Siyah görsel


backbone_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")


try:
    model = AestheticScorer(backbone=backbone_model)
except AttributeError:
    model = AestheticScorer(backbone=backbone_model.vision_model)

state_dict = torch.load("model.pt", map_location=torch.device('cpu'))
model.load_state_dict(state_dict)
model.eval()  


processor = CLIPProcessor.from_pretrained("rsinema/aesthetic-scorer")
inputs1 = processor(images=image1, return_tensors="pt")["pixel_values"]
inputs2 = processor(images=image2, return_tensors="pt")["pixel_values"]
inputs3 = processor(images=image3, return_tensors="pt")["pixel_values"]
inputs4 = processor(images=image4, return_tensors="pt")["pixel_values"]
inputs_black = processor(images=image_black, return_tensors="pt")["pixel_values"]


with torch.no_grad():
    scores1 = model(inputs1)
    scores2 = model(inputs2)
    scores3 = model(inputs3)
    scores4 = model(inputs4)
    scores_black = model(inputs_black)


aesthetic_categories = ["Overall", "Quality", "Composition", "Lighting", "Color", "Depth of Field", "Content"]

print("Normal Görsel 1 Skorları:")
for category, score in zip(aesthetic_categories, scores1):
    print(f"{category}: {score.item():.2f}/5")

print("\nNormal Görsel 2 Skorları:")
for category, score in zip(aesthetic_categories, scores2):
    print(f"{category}: {score.item():.2f}/5")

print("\nNormal Görsel 3 Skorları:")
for category, score in zip(aesthetic_categories, scores3):
    print(f"{category}: {score.item():.2f}/5")

print("\nNormal Görsel 4 Skorları:")
for category, score in zip(aesthetic_categories, scores4):
    print(f"{category}: {score.item():.2f}/5")

print("\nSiyah Görsel Skorları:")
for category, score in zip(aesthetic_categories, scores_black):
    print(f"{category}: {score.item():.2f}/5")
```
